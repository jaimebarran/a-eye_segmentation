```python
''' N=1157 dataframe'''

import pandas as pd
import os

csv = '/mnt/sda1/Repos/a-eye/Output/axial_length/nnunet/axial_length_nnunet_3D_N1157_left_eyes.csv'
df = pd.read_csv(csv)

images_path = "/mnt/sda1/Repos/a-eye/Data/SHIP_dataset/non_labeled_dataset/evaluation_left_eyes/images"

# Get image filenames from directory
image_files = os.listdir(images_path)
image_subjects = set(int(f.split('.')[0]) for f in image_files)

# Get subject IDs from dataframe
df_subjects = set(df['Subject'].values)

# Check which image subjects are in dataframe
missing_subjects = image_subjects - df_subjects
in_dataframe = image_subjects & df_subjects

print(f"Images: {len(image_subjects)}, DataFrame subjects: {len(df_subjects)}")
print(f"Images in dataframe: {len(in_dataframe)}")
print(f"Images NOT in dataframe: {len(missing_subjects)}")
if missing_subjects:
    print(f"Missing subjects: {sorted(missing_subjects)}")
```
